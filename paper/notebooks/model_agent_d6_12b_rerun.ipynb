# ATLAS d6 12B full batch, rerun with artifact packing

Same 22-task sample, same 12B checkpoint, vLLM 0.32.0, max_model_len 6144.
This rerun exists to retrieve the per-task patches: the previous run's outputs
could not be downloaded because Kaggle truncates large outputs and the workspace
trees are hundreds of MB. The batch now packs summaries and diffs into
artifacts.zip (a few hundred KB) at the end, so one download retrieves all evidence.

The runner also carries two fixes: prompt-token accounting reads the request-level
result, and the prompt is budgeted against the real chat template before inference.
Those changes make token counts trustworthy but also mean this run is not
token-identical to the previous 12B batch; patch outcomes remain comparable.
