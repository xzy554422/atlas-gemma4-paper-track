# ATLAS: bounded Gemma 4 E4B runtime probe (v2, SM 7.5 tile patch)

v1 failed before the first token on Tesla T4: Triton unified attention requested
98,304 B of shared memory against the 65,536 B hardware limit (vLLM issue
#38918). v2 applies the community-verified workaround before engine start: on
compute capability < 8.0, `_get_tile_size` falls back to tile 16 when
`head_size > 128`. Everything else stays bounded: one attempt, one model
variant, FP16, tensor parallelism 1, 2,048-token context, 32 generated tokens,
subprocess with external timeout. This probe establishes generation
compatibility only; it never measures agent repair performance.
